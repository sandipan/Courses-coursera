In this [paper](https://www.microsoft.com/en-us/research/wp-content/uploads/2016/02/Ferreira_Fisher_Sample_Oriented_Tasks.pdf) the authors describe the challenges users face when trying to make judgements about probabilistic data generated through samples. As an example, they look at a bar chart of four years of data (replicated below in Figure 1). Each year has a y-axis value, which is derived from a sample of a larger dataset. For instance, the first value might be the number votes in a given district or riding for 1992, with the average being around 33,000. On top of this is plotted the 95% confidence interval for the mean (see the boxplot lectures for more information, and the yerr parameter of barcharts).

A challenge that users face is that, for a given y-axis value (e.g. 42,000), it is difficult to know which x-axis values are most likely to be representative, because the confidence levels overlap and their distributions are different (the lengths of the confidence interval bars are unequal). One of the solutions the authors propose for this problem (Figure 2c) is to allow users to indicate the y-axis value of interest (e.g. 42,000) and then draw a horizontal line and color bars based on this value. So bars might be colored red if they are definitely above this value (given the confidence interval), blue if they are definitely below this value, or white if they contain this value.

```python
# Use the following data for this assignment:

import pandas as pd
import numpy as np

np.random.seed(12345)

df = pd.DataFrame([np.random.normal(32000,200000,3650), 
                   np.random.normal(43000,100000,3650), 
                   np.random.normal(43500,140000,3650), 
                   np.random.normal(48000,70000,3650)], 
                  index=[1992,1993,1994,1995])
df
```

```python
#Your Code Here
import matplotlib.pylab as plt

df['m'] = df.mean(axis=1)
df['s'] = df.std(axis=1)
n = df.shape[1]

df['ci'] = 1.96*df['s']/np.sqrt(n)
df_ci = df[['m', 'ci']]
df_ci.head()

plt.figure(figsize=(6,6))
plt.bar(x=df_ci.index, height=df_ci.m, yerr=df_ci.ci, width=1.0, color='lightgray', edgecolor='k', capsize=5)
plt.xticks(df_ci.index)
plt.show()
```

```python
#Your Code Here
#%matplotlib notebook
%matplotlib ipympl
import matplotlib.pylab as plt
from matplotlib.colors import LinearSegmentedColormap
from ipywidgets import *
from matplotlib.artist import Artist

colors = ['#68001e', '#d75f49', '#f7f7f7', '#3f92c5', '#022f63'] #reversed(['#022f63', '#3f92c5', "#f7f7f7", '#d75f49', '#68001e'])
positions = [0, 0.25, 0.5, 0.75, 1]
N = 100
cmap = LinearSegmentedColormap.from_list('rb_colormap', list(zip(positions, colors)), N)
colors = [matplotlib.colors.to_hex(cmap(i)) for i in range(N)]
cmap

df['m'] = df.mean(axis=1)
df['s'] = df.std(axis=1)
n = df.shape[1]

df['ci'] = 1.96*df['s']/np.sqrt(n)
df_ci = df[['m', 'ci']]

y = 40000
df_ci['color'] = 'gray'
for i in range(len(df_ci)):
    m, ci = df_ci.iloc[i,0], df_ci.iloc[i,1]
    lci, uci = m - ci, m + ci
    index = 0 if y < lci else 99 if y > uci else int((100* (y - lci) / (2*ci))) #np.round
    df_ci.at[df_ci.index[i], 'color'] = colors[index]    

fig = plt.figure('Plotting probabilities with matplotlib', figsize=(6,6))
plt.bar(x=df_ci.index, height=df_ci.m, yerr=df_ci.ci, width=1.0, color=df_ci.color, edgecolor='k', capsize=5)
line = plt.axhline(y = y, color = 'gray', linestyle = '-') 
text = plt.text(plt.xlim()[0]+0.35, y+1000, 'y={}'.format(y), fontsize=10, va='center', ha='center')
Artist.set_visible(text, True)
plt.xticks(df_ci.index)
plt.title('Sample-Oriented Task-Driven Visualizations with matplotlib', size=12)
plt.xlabel('year')
plt.ylabel('probability')

def update(y = 40000):
    global line, text
    for i in range(len(df_ci)):
        m, ci = df_ci.iloc[i,0], df_ci.iloc[i,1]
        lci, uci = m - ci, m + ci
        index = 0 if y < lci else 99 if y > uci else int(100 * (y - lci) / (2*ci))
        #print(index)
        df_ci.at[df_ci.index[i], 'color'] = colors[index]    
    plt.bar(x=df_ci.index, height=df_ci.m, yerr=df_ci.ci, width=1.0, color=df_ci.color, edgecolor='k', capsize=5)
    line.remove()
    Artist.remove(text)
    line = plt.axhline(y = y, color = 'gray', linestyle = '-') 
    text = plt.text(plt.xlim()[0]+0.35, y+1000, 'y={}'.format(y), fontsize=10, va='center', ha='center')
    Artist.set_visible(text, True)
    plt.xticks(df_ci.index)
    plt.title('Sample-Oriented Task-Driven Visualizations with matplotlib', size=12)
    plt.xlabel('year')
    plt.ylabel('probability')
    fig.canvas.draw_idle()
    
interact(update, y = (20000,50000,100));
```

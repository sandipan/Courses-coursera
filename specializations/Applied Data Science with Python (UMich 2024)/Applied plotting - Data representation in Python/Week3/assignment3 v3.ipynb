In this [paper](https://www.microsoft.com/en-us/research/wp-content/uploads/2016/02/Ferreira_Fisher_Sample_Oriented_Tasks.pdf) the authors describe the challenges users face when trying to make judgements about probabilistic data generated through samples. As an example, they look at a bar chart of four years of data (replicated below in Figure 1). Each year has a y-axis value, which is derived from a sample of a larger dataset. For instance, the first value might be the number votes in a given district or riding for 1992, with the average being around 33,000. On top of this is plotted the 95% confidence interval for the mean (see the boxplot lectures for more information, and the yerr parameter of barcharts).

A challenge that users face is that, for a given y-axis value (e.g. 42,000), it is difficult to know which x-axis values are most likely to be representative, because the confidence levels overlap and their distributions are different (the lengths of the confidence interval bars are unequal). One of the solutions the authors propose for this problem (Figure 2c) is to allow users to indicate the y-axis value of interest (e.g. 42,000) and then draw a horizontal line and color bars based on this value. So bars might be colored red if they are definitely above this value (given the confidence interval), blue if they are definitely below this value, or white if they contain this value.

```python
# Use the following data for this assignment:
import pandas as pd
import numpy as np

np.random.seed(12345)

df = pd.DataFrame([np.random.normal(32000,200000,3650), 
                   np.random.normal(43000,100000,3650), 
                   np.random.normal(43500,140000,3650), 
                   np.random.normal(48000,70000,3650)], 
                  index=[1992,1993,1994,1995])
df
```

```python
%matplotlib ipympl

import matplotlib.pyplot as plt
import matplotlib.cm as cm
from ipywidgets import interact # This is a widget for jupyter 
import ipywidgets as widgets
# This module enable pandas to communicate with matplotlib
from pandas.plotting import register_matplotlib_converters
register_matplotlib_converters()

# Function for finding the % of overlapping area of y-band and confidence interval of each bar
def find_overlap(df, region_range):
    overlap_percent=0
    if (df['temp_upper']-df['temp_lower'])<0:
        return overlap_percent
    else:
        overlap_percent=(df['temp_upper']-df['temp_lower'])/(region_range)
        if overlap_percent>1:
            overlap_percent=1
        return overlap_percent
       
def plot_chart(y_lower, y_upper):
    global line_lower, line_upper, cbar
    
    # Clear chart
    plt.cla()

    # Get the 'Reds' colormap
    cmap = plt.get_cmap('Reds', 9)

    # Get the data for the plot. 
    xvals=[str(x) for x in df.index]
    y_stats=pd.DataFrame()
    y_stats['mean']=df.mean(axis=1)

    # Calculate the confidence interval
    error_array=1.96*df.std(axis=1)/np.sqrt(df.count(axis=1))
    
    # Find out the overlapping area (if any)
    y_stats['temp_upper']=[x if x < y_upper else y_upper for x in (y_stats['mean']+error_array)]
    y_stats['temp_lower']=[x if x > y_lower else y_lower for x in (y_stats['mean']-error_array)]
    y_stats['overlap']=y_stats[['temp_lower', 'temp_upper']].apply(lambda x: find_overlap(x, y_upper-y_lower), axis=1)    

    # Draw the bar chart
    # By calling cmap(y_stats['overlap']), it will map the y_starts['overlap'] value 
    # (which is in the range of [0, 1] into a colour using colourmap 'Reds'
    plt.bar(xvals, y_stats['mean'], width=1, color=cmap(y_stats['overlap']), edgecolor='black')
    plt.errorbar(xvals, y_stats['mean'], yerr=error_array, color='black', fmt=' ' , capsize=20, capthick=3)
    
    # Draw back the y-band
    # The transform=ax.transAxes parameter changes the coordinate reference system to axes
    line_lower=ax.axhline(y=y_lower, color='b', linestyle='--')
    plt.text(0.32, y_lower/60000+0.01, f'y={str(round(y_lower))}', transform=ax.transAxes, color='b')
    line_upper=ax.axhline(y=y_upper, color='g', linestyle='--')    
    plt.text(0.55, y_upper/60000+0.01, f'y={str(round(y_upper))}', transform=ax.transAxes, color='g')

    plt.gcf().suptitle('Compare different sample means with a y-band', y=0.94, fontsize=16)
    ax.set_ylim(0, 60000)
    # The transform=plt.gcf().transFigure parameter changes the coordinate reference system to whole subplot
    plt.text(x=0.13, y=0.04, s='Instructions: (1) Click and set the 1st line. Keep holding your mouse click.', transform=plt.gcf().transFigure)
    plt.text(x=0.245, y=0.01, s='(2) Move your mouse and release the click to set the 2nd line.',  transform=plt.gcf().transFigure)
    
    # Draw the colour bar
    if cbar is None:
        sm = cm.ScalarMappable(cmap=cmap)
        cbar = plt.colorbar(sm, ticks=np.linspace(0, 1, 10), ax=ax)
        cbar.set_label('Probabilities')

    plt.draw()

def on_click(event):
    global line_lower, line_upper, line_y_pos
    if event.inaxes==ax:
        line_y_pos[0]=event.ydata
        if line_lower:           
            line_lower.set_ydata([line_y_pos[0]])
        else:
            line_lower=ax.axhline(y=line_y_pos[0], color='b', linestyle='--')
        plt.draw()

def on_release(event):
    global line_lower, line_upper, line_y_pos
    if event.inaxes==ax:
        line_y_pos[1]=event.ydata
        if line_y_pos[0]>line_y_pos[1]:
            plot_chart(line_y_pos[1],line_y_pos[0])
        else:
            plot_chart(line_y_pos[0],line_y_pos[1])        

line_y_pos=[30000,40000]
line_lower=None
line_upper=None
cbar=None

fig, ax=plt.subplots(figsize=[8,6])
plot_chart(line_y_pos[0],line_y_pos[1])
fig.canvas.mpl_connect('button_press_event', on_click)
fig.canvas.mpl_connect('button_release_event', on_release)
plt.show()
```

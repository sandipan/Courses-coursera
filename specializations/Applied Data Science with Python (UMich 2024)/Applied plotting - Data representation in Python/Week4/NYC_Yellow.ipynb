### NYC dataset

```python
# https://www.nyc.gov/site/tlc/about/tlc-trip-record-data.page
import pandas as pd
df_traffic = pd.read_parquet('data/yellow_tripdata_2023-04.parquet', engine='fastparquet')
df_traffic.to_csv('data/yellow_tripdata_2023-04.csv')
df_traffic.head()
```

```python
df_PU = df_traffic.groupby('PULocationID').agg({'passenger_count': 'sum'}).reset_index()
df_DO = df_traffic.groupby('DOLocationID').agg({'passenger_count': 'sum'}).reset_index()
df_PU.head()
```

```python
import geopandas as gpd
df_map = gpd.read_file("taxi_zones/taxi_zones.shp")
df_map.plot()
```

```python
df_wm = df_map.to_crs(epsg=3857)
```

# Random Number Generation and Histograms

The goal of a random number generator is to produce realizations (numerical observations) of a continuous uniform random variable on the interval $(0,1)$. In this course, we will look at many algorithms that will take these values and turn them in to much more interesting things! The built-in random number generator will be one of the only places where we use a "black box" to "magically" produce something.

<br>
The code is easy enough. In order to produce $n$ realizations from the continuous uniform distribution on $(0,1)$, we would type <code>np.random.rand(n)</code>.  Let's try it.

```python
# Run this cell
np.random.rand(5)
```

Suppose that $U \sim unif(0,1)$. The probability that $U$ is, for example, between $0.14$ and $0.57$ is $0.57-0.14=0.43$. If we were to simulate/generate many values with a random number generator, we should see that the proportion of values in this interval is approximately $0.43$. Let's try it.

```python
# simulate 100,000 values from the unif(0,1) distribution
mysample = np.random.rand(100000)     

# find the length of the mysample vector for which the condition is satisfied
# and divide by the total length of the mysample vector
len(mysample[(mysample > 0.14) & (mysample < 0.57)])/len(mysample)
```

Due to sampling variability, we do not expect to see exactly $0.43$ here. We do, however, expect to see a value between

<br>
$$
0.43 - 2 \sqrt{\frac{0.43*(1-0.43)}{100000}} \approx 0.426869
$$

and 

$$
0.43 + 2 \sqrt{\frac{0.43*(1-0.43)}{100000}} \approx 0.433131
$$

roughly 95% of the time. Justification for this range of values is given in a course on statistical inference. Do not be concerned if these expressions look unfamiliar to you. Understanding them will not affect your ability to do well in this course!

If the proportion you just computed does not fall between these values, try running the previous cell again!

We will now make a histogram of "mysample". While we could do this by simply typing <code>plt.hist(mysample)</code>, this would put counts for each bin on the y-axis. As discussed in a previous video, we would like to have "density" on the y-axis. This is achieved by adding <code>density='True'</code> to the plt.hist command.

```python
# Run this cell
plt.hist(mysample,density='True')        # make the histogram
plt.show()                               # show the histogram (Your histogram will show without this line
                                         # but it is a good habit to get into for the ultimate display 
                                         # of more complex plots and you will get a nicer looking
                                         # plot even here!)
```

That is one ugly histogram! Let's clean it up a bit by making the edges more well-defined.

```python
# Run this cell
plt.hist(mysample,edgecolor='black', density='True')      # make the histogram
plt.show()                                                # show the histogram
```

While the default bin with chosen by Python here actually looks pretty good, we should know how to change it. For more complex-shaped distributions, a bin width that is too large will result in a loss of detail in the shape of the histogram while a bin width that is too small will result in a very "spikey" result, often with artificially empty bins between bins containing observations. Just for fun, let's change the bin width to $0.05$. We can do this by defining a sequence of break points between $0$ and $1$ as in the following cell.

```python
# Run this cell
br = np.arange(0.0, 1.05, 0.05)                                    # define the bin breaks
plt.hist(mysample, bins=br, edgecolor='black', density='True')     # make the histogram
plt.show()                                                         # show the histogram
```

Let's return to the original bin width of $0.1$ but have it controlled by us. That is, we will explicity define the bin break points. As a final touch, we'll change the color of the bars and put the uniform (0,1) flat line pdf at height 1. This line can be defined by two points. We'll connect the (𝑥,𝑦) pairs (0,1) and (1,1). We will do this with the "plt.plot" function that will take an array of 𝑥-coordinates and an array of 𝑦-coordinates for the points to be connected. In this case, the 𝑥-coordinates are 0 and 1 and the 𝑦-coordinates are 1 and 1 so it may be a little confusing!

```python
# Run this cell
br = np.arange(0.0, 1.1, 0.1)                                                       # define the bin breaks
plt.hist(mysample, bins=br, color='skyblue', edgecolor='black', density='True')     # make the histogram
plt.plot([0,1],[1,1],color='black')                                                 # line through points, list both x-coordinates then both y-coordinates!
plt.show()                                                                          # show the histogram
```

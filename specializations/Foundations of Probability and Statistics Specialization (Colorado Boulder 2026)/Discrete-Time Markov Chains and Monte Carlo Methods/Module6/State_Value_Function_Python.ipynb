# The State Value Function Computation
This is not an assignment. It is a notebook containing the state value function computation for Module 6, Lesson 4. Recall that we had a maze or "grid" consisting of 14 cells. 

<br>
<img src="state_value_grid.png" width="180"> <br>

We want to guide an agent moving around the grid, continuously in time, with the goal of maximize expected cumulative rewards over time. The rules are as follows.
<ul>
<li> If the agent is in any numbered cell, other than cell 10, the agent will attempt to move either up, down, left, or right, with probability 1/4 each.</li>
    <li> If the agent moves into any numbered cell, including cell 10, the agent receives a reward of 0.</li>
    <li> If, when moving up, down, left, or right, the agent attempts to move off of the grid, the agent will stay in the current position and receive a reward of -1.</li>
    <li> The blue cells are blocked. If the agent attempts to move into a blocked cell the agent will stay in the current position and receive a reward of -1.</li>
        
<li> If the agent moves into cell 10, in the next move, the agent is sent to cell 6 with probability 1 and will collect a reward of +5.</li>
</ul>

We will now compute the state value function using a discount factor of $\gamma=0.9$. Recall that the formula is

$$
\vec{v}_{\pi} = (I-\gamma {\bf{P}}_{\pi})^{-1} \, \vec{r}_{\pi}.
$$

```python
# Import the NumPy library
import numpy as np
```

```python
# Set the discount factor
gamma = 0.9

# Define the rewards vector (See the video for a detailed explanation of these values!)
r = np.array([-0.5, -0.25, -0.5, -0.5, -0.25, -0.5, -0.5, -0.5, -0.5, 5, -0.5, -0.5, -0.25, -0.5])
```

```python
# Define the P matrix (See the video for a detailed explanation of these values!)
P = np.zeros((14,14))
P[0,[0,1,4]] = np.array([0.5,0.25,0.25])   # Recall the zero-based indexing, this is setting the (1,1), (1,2), 
                                           # and (1,5) entries of P
P[1,[0,1,2,5]] = 0.25
P[2,[1,2,3]] = np.array([0.25,0.5,0.25])
P[3,[2,3,6]] = np.array([0.25,0.5,0.25])
P[4,[0,4,5,7]] = 0.25
P[5,[1,4,5]] = np.array([0.25,0.25,0.5])
P[6,[3,6,9]] = np.array([0.25,0.5,0.25])
P[7,[4,7,10]] = np.array([0.25,0.5,0.25])
P[8,[8,9,12]] = np.array([0.5,0.25,0.25])
P[9,5] = 1
P[10,[7,10,11]] = np.array([0.25,0.5,0.25])
P[11,[10,11,12]] = np.array([0.25,0.5,0.25])
P[12,[8,11,12,13]] = 0.25
P[13,[9,12,13]] = np.array([0.25,0.25,0.5])
```

```python
P
```

```python
# Define a 14x14 identity matrix
ident = np.identity(14)
```

```python
# Compute the state value function
np.matmul(np.linalg.inv(ident-gamma*P),r)
```

Below, we review four standard types of selections. To visualize them, we use a toy example of selecting k=2k=2 objects out of a set \{\text{a}, \text{b}, \text{c}\}{a,b,c} of size n=3n=3. We distinguish ordered and unordered selections. In mathematics, the ordered ones are written in parentheses (that is, round brackets) whereas the unordered ones are written in braces (that is, curly brackets): we treat (\text{c}, \text{a})(c,a) and (\text{a}, \text{c})(a,c) as different, however we consider \{\text{c}, \text{a}\}{c,a} and \{\text{a}, \text{c}\}{a,c} as being the same. We also distinguish selections with and without repetitions: (\text{c}, \text{c})(c,c) contains a repeating element, whereas (\text{c}, \text{a})(c,a) does not.

```python
from statistics import mean


def improve_average(numbers):
    minimal, average = min(numbers), mean(numbers)
    while minimal < average:
        print(f'The mean of {numbers} is {average:.2f}')
        print(f'Remove {minimal}')
        numbers.remove(minimal)
        minimal, average = min(numbers), mean(numbers)


improve_average([6, 9, 5, 2, 4, 3])
```
